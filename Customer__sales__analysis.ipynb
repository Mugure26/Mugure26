```python
import pandas as pd

data = {
    "Date": [
        "2025-01-05", "2025-01-12", "2025-02-03", "2025-02-15",
        "2025-03-02", "2025-03-18", "2025-04-07", "2025-04-21",
        "2025-05-04", "2025-05-19", "2025-06-08", "2025-06-25",
        "2025-07-10", "2025-07-28", "2025-08-05", "2025-08-20",
        "2025-09-03", "2025-09-17", "2025-10-06", "2025-10-22"
    ],
    "Product": [
        "Laptop", "Mouse", "Keyboard", "Laptop",
        "Monitor", "Mouse", "Keyboard", "Laptop",
        "Monitor", "Headphones", "Laptop", "Mouse",
        "Keyboard", "Monitor", "Headphones", "Laptop",
        "Mouse", "Keyboard", "Laptop", "Monitor"
    ],
    "Category": [
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Accessories", "Electronics", "Accessories",
        "Accessories", "Electronics", "Accessories", "Electronics",
        "Accessories", "Accessories", "Electronics", "Electronics"
    ],
    "Region": [
        "Nairobi", "Mombasa", "Nairobi", "Kisumu",
        "Nakuru", "Nairobi", "Mombasa", "Kisumu",
        "Nairobi", "Nakuru", "Mombasa", "Nairobi",
        "Kisumu", "Nakuru", "Nairobi", "Mombasa",
        "Kisumu", "Nairobi", "Nakuru", "Mombasa"
    ],
    "Quantity": [
        2, 10, 5, 3, 4, 15, 8, 2,
        5, 7, 3, 12, 6, 4, 10, 2,
        20, 9, 3, 6
    ],
    "Unit_Price": [
        85000, 1500, 3500, 85000, 25000, 1500, 3500, 85000,
        25000, 5000, 85000, 1500, 3500, 25000, 5000, 85000,
        1500, 3500, 85000, 25000
    ]
}

df = pd.DataFrame(data)

df["Date"] = pd.to_datetime(df["Date"])
df["Revenue"] = df["Quantity"] * df["Unit_Price"]

df.head()
```

```python
df["Revenue"] = df["Quantity"] * df["Unit_Price"]
```

```python
print("Number of rows and columns:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python
df.describe()
```

visualization


```python
product_revenue = df.groupby("Product")["Revenue"].sum().sort_values(ascending=False)

print(product_revenue)
```

```python
plt.figure(figsize=(10, 6))

product_revenue.plot(kind="bar")

plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue (KSh)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
print(product_revenue)
```

analysis of quantity sold


```python
product_quantity = df.groupby("Product")["Quantity"].sum().sort_values(ascending=False)

print(product_quantity)
```

```python
plt.figure(figsize=(10, 6))

product_quantity.plot(kind="bar")

plt.title("Quantity Sold by Product")
plt.xlabel("Product")
plt.ylabel("Quantity Sold")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

ANAYLSIS BY REGION


```python
region_revenue = df.groupby("Region")["Revenue"].sum().sort_values(ascending=True)

print(region_revenue)
```

sales over time


```python
df["Month"] = df["Date"].dt.to_period("M").astype(str)

monthly_revenue = df.groupby("Month")["Revenue"].sum()

print(monthly_revenue)
```

```python
plt.figure(figsize=(12, 6))

monthly_revenue.plot(kind="line", marker="o")

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (KSh)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
print(monthly_revenue)
```

```python
df["Month"] = df["Date"].dt.month

monthly_revenue = df.groupby("Month")["Revenue"].sum()

# Add missing months
monthly_revenue = monthly_revenue.reindex(range(1, 13), fill_value=0)

# Give months proper names
month_names = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

monthly_revenue.index = month_names

print(monthly_revenue)
```

```python
plt.figure(figsize=(12, 6))

monthly_revenue.plot(kind="line", marker="o")

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (KSh)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
region_product = df.pivot_table(
    index="Region",
    columns="Product",
    values="Revenue",
    aggfunc="sum",
    fill_value=0
)

print(region_product)
```

```python
plt.figure(figsize=(12, 6))

sns.heatmap(region_product, annot=True, fmt=".0f")

plt.title("Revenue by Region and Product")
plt.xlabel("Product")
plt.ylabel("Region")

plt.tight_layout()
plt.show()
```

visualization by region

```python
region_product.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Revenue by Region and Product")
plt.xlabel("Region")
plt.ylabel("Revenue (KSh)")
plt.xticks(rotation=0)
plt.legend(title="Product")

plt.tight_layout()
plt.show()
```

correlation analysis:investigate whether Quantity, Unit Price, and Revenue are related.

```python
correlation = df[["Quantity", "Unit_Price", "Revenue"]].corr()

print(correlation)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Between Sales Variables")

plt.tight_layout()
plt.show()
```

```python
total_revenue = df["Revenue"].sum()
total_units = df["Quantity"].sum()
total_transactions = len(df)
average_transaction = df["Revenue"].mean()

print("===== SALES KPI SUMMARY =====")
print(f"Total Revenue: KSh {total_revenue:,.0f}")
print(f"Total Units Sold: {total_units:,}")
print(f"Total Transactions: {total_transactions:,}")
print(f"Average Transaction Value: KSh {average_transaction:,.0f}")
```

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

kpis = [
    ("Total Revenue", f"KSh {total_revenue:,.0f}"),
    ("Units Sold", f"{total_units:,}"),
    ("Transactions", f"{total_transactions:,}"),
    ("Average Sale", f"KSh {average_transaction:,.0f}")
]

for ax, (title, value) in zip(axes, kpis):
    ax.text(
        0.5, 0.5, value,
        ha="center",
        va="center",
        fontsize=20,
        fontweight="bold"
    )
    ax.set_title(title)
    ax.axis("off")

plt.suptitle("Customer Sales Analysis — KPI Dashboard", fontsize=18)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Create a 2x2 dashboard
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# 1. Revenue by Product
product_revenue = df.groupby("Product")["Revenue"].sum().sort_values()

axes[0, 0].barh(product_revenue.index, product_revenue.values)
axes[0, 0].set_title("Revenue by Product")
axes[0, 0].set_xlabel("Revenue (KSh)")

# 2. Revenue by Region
region_revenue = df.groupby("Region")["Revenue"].sum().sort_values()

axes[0, 1].bar(region_revenue.index, region_revenue.values)
axes[0, 1].set_title("Revenue by Region")
axes[0, 1].set_ylabel("Revenue (KSh)")

# 3. Monthly Revenue
monthly_revenue = df.groupby("Month")["Revenue"].sum()

axes[1, 0].plot(monthly_revenue.index.astype(str),
                monthly_revenue.values,
                marker="o")

axes[1, 0].set_title("Monthly Revenue Trend")
axes[1, 0].set_xlabel("Month")
axes[1, 0].set_ylabel("Revenue (KSh)")
axes[1, 0].tick_params(axis="x", rotation=45)

# 4. Revenue by Category
category_revenue = df.groupby("Category")["Revenue"].sum()

axes[1, 1].bar(category_revenue.index, category_revenue.values)
axes[1, 1].set_title("Revenue by Category")
axes[1, 1].set_ylabel("Revenue (KSh)")

plt.suptitle(
    "Customer Sales Analysis — Business Insights Dashboard",
    fontsize=18
)

plt.tight_layout()
plt.show()
```
