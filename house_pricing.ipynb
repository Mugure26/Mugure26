prediction model training:one_hot encoding for zipcode

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

categorical_features = ["zipcode"]

numeric_features = [col for col in X.columns if col != "zipcode"]

preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

Train Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline

linear_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])

linear_model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")
```

make prediction

```python
y_pred_linear = linear_model.predict(X_test)

print("Predictions made successfully!")
print("First 5 predictions:")
print(y_pred_linear[:5])
```

evaluate linear regression

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

r2_linear = r2_score(y_test, y_pred_linear)
mae_linear = mean_absolute_error(y_test, y_pred_linear)
rmse_linear = np.sqrt(mean_squared_error(y_test, y_pred_linear))

print("Linear Regression Results")
print("-------------------------")
print("R² Score:", r2_linear)
print("MAE:", mae_linear)
print("RMSE:", rmse_linear)
```

train the Decision Tree

```python
from sklearn.tree import DecisionTreeRegressor

tree_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", DecisionTreeRegressor(
        random_state=42,
        max_depth=20
    ))
])

tree_model.fit(X_train, y_train)

print("Decision Tree model trained successfully!")
```

prediction with decision tree

```python
y_pred_tree = tree_model.predict(X_test)

print("Decision Tree predictions made successfully!")
print("First 5 predictions:")
print(y_pred_tree[:5])
```

```python
evaluate prediction
```

```python
r2_tree = r2_score(y_test, y_pred_tree)
mae_tree = mean_absolute_error(y_test, y_pred_tree)
rmse_tree = np.sqrt(mean_squared_error(y_test, y_pred_tree))

print("Decision Tree Results")
print("--------------------")
print("R² Score:", r2_tree)
print("MAE:", mae_tree)
print("RMSE:", rmse_tree)
```

random forest


```python
from sklearn.ensemble import RandomForestRegressor

forest_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=100,
        max_depth=20,
        random_state=42,
        n_jobs=-1
    ))
])

forest_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

random forest prediction

```python
y_pred_forest = forest_model.predict(X_test)

print("Random Forest predictions made successfully!")
print("First 5 predictions:")
print(y_pred_forest[:5])
```

evalute random forest

```python
r2_forest = r2_score(y_test, y_pred_forest)
mae_forest = mean_absolute_error(y_test, y_pred_forest)
rmse_forest = np.sqrt(mean_squared_error(y_test, y_pred_forest))

print("Random Forest Results")
print("--------------------")
print("R² Score:", r2_forest)
print("MAE:", mae_forest)
print("RMSE:", rmse_forest)
```

Which features are most important?

```python
# Get the trained Random Forest model
rf_model = forest_model.named_steps["model"]

# Get the preprocessing step
preprocessor_fitted = forest_model.named_steps["preprocessor"]

# Get feature names after preprocessing
feature_names = preprocessor_fitted.get_feature_names_out()

# Get feature importance
importances = rf_model.feature_importances_

# Create a DataFrame
feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importances
}).sort_values(by="Importance", ascending=False)

print(feature_importance.head(15))
```
